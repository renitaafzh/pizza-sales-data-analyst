```python
#import library
import pandas as pd

# Load All data Raws
orders=pd.read_csv("orders.csv")
order_details =pd.read_csv("order_details.csv")
pizzas=pd.read_csv("pizzas.csv")
pizza_types= pd.read_csv("pizza_types.csv", encoding="latin1")
```

```python
# Store all df in a dictionary
datasets = {
    "orders": orders,
    "order_details": order_details,
    "pizza_types": pizza_types,
    "pizzas": pizzas
}

# Loop through each dataset to inspect structure, missing values, and duplicates
for name, df in datasets.items():
    print(f"== {name.upper()} ==")
    df.info()
    print("Missing values:\n", df.isnull().sum())
    print("Duplicates:", df.duplicated().sum(), "\n")
```

```python
# Convert data collumn and extract name dayname & order hour
orders['date']=pd.to_datetime(orders['date'])
orders['dayname']=orders['date'].dt.day_name()
orders['order_hour']= pd.to_datetime(orders['time'], format='%H:%M:%S').dt.hour
```

```python
# Strip extra white case from text collumn
pizzas['size']= pizzas['size'].str.strip()
pizza_types['name']=pizza_types['name'].str.strip()
pizza_types['category']=pizza_types['category'].str.strip()

orders.head()
```

```python
#Export cleaned datasets to CSV
orders.to_csv("orders_clean.csv", index=False, encoding="utf8")
order_details.to_csv("order_details_clean.csv", index=False, encoding="utf8")
pizzas.to_csv("pizzas_clean.csv", index=False, encoding="utf8")
pizza_types.to_csv('pizza_types_clean.csv', index=False, encoding='utf')

print('All cleaned files exported successfully!')
```
